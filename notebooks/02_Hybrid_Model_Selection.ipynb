## 1. Research Direction and Scope

### 1.1 Research Title

**Diagnostic-Driven Hybrid Machine Learning for ECG Analysis**

Authors: Haris Md Sahed and Fatih Ozaydin.

### 1.2 Research Background

The first experimental stage investigated the statistical and signal-level characteristics of the MIT-BIH Arrhythmia Database.

The initial analysis examined stationarity, rolling statistics, feature correlations, outliers, and differences between recording-level statistical groups.

The findings suggest that ECG recordings may exhibit different statistical characteristics. These observations motivate an investigation into whether dataset characteristics can inform the selection of appropriate machine learning strategies.

### 1.3 Research Problem

Conventional ECG classification approaches often apply predefined machine learning or deep learning models without systematically considering the statistical and signal-level characteristics of the input dataset.

However, differences in signal variability, temporal behaviour, class distribution, and recording conditions may influence the suitability and performance of different modelling strategies.

This research investigates whether a diagnostic-driven approach can use dataset characteristics to guide the selection of appropriate machine learning and hybrid modelling strategies for ECG classification.

### 1.4 Central Research Idea

The central idea is to investigate a diagnostic-driven modelling framework that follows four stages:

1. Examine the statistical and signal-level characteristics of the ECG dataset.
2. Identify potential modelling challenges using diagnostic information.
3. Select appropriate candidate preprocessing and modelling strategies.
4. Evaluate the selected strategies through controlled experiments.

The framework aims to investigate whether diagnostic information can provide useful guidance beyond selecting a single predefined model for every dataset.

### 1.5 Primary Research Contribution

The primary contribution of this research is the investigation and experimental evaluation of a diagnostic-driven model selection approach for ECG classification.

Rather than proposing only another ECG classification architecture, the research focuses on the relationship between dataset characteristics, modelling strategy selection, and classification performance.

### 1.6 Research Scope

The initial experiments will use the MIT-BIH Arrhythmia Database.

The investigation will consider:

- Statistical and signal-level diagnostics.
- Traditional machine learning approaches.
- Deep learning approaches.
- Hybrid modelling strategies.
- Class imbalance.
- Model performance and robustness.
- Diagnostic-driven model selection.

The initial experiments will focus on heartbeat-level ECG classification while retaining recording-level diagnostic information.

### 1.7 Research Limitations at This Stage

The statistical findings obtained during the first experimental stage are exploratory.

The observed differences between KPSS groups do not establish clinical significance or predictive value.

The usefulness of diagnostic features for model selection must be investigated through independent experimental evaluation.

No preprocessing, model training, or predictive evaluation has been conducted at this stage.

## 3. Candidate Modelling Strategies

### 3.1 Overview

This research will investigate three main modelling families:

1. Traditional machine learning.
2. Deep learning.
3. Hybrid machine learning and deep learning.

The candidate models are selected based on their different learning mechanisms and their potential suitability for ECG heartbeat classification.

### 3.2 Traditional Machine Learning

#### Random Forest (RF)

Random Forest is an ensemble learning algorithm that combines multiple decision trees.

It will be investigated as a traditional baseline using engineered ECG features, including morphological, statistical, and RR interval characteristics.

**Advantages:**
- Captures nonlinear feature relationships.
- Provides feature importance estimates.
- Requires relatively limited preprocessing compared with some other methods.

**Limitations:**
- Depends on the quality of engineered features.
- May favour majority classes under severe class imbalance.

#### Support Vector Machine (SVM)

SVM identifies decision boundaries that separate different heartbeat classes.

It will be investigated using appropriately scaled engineered ECG features.

**Advantages:**
- Can model nonlinear classification boundaries using kernel functions.
- Suitable for structured feature representations.

**Limitations:**
- Requires parameter and kernel selection.
- Computational requirements can increase with dataset size.
- Sensitive to feature scaling and class imbalance.

#### XGBoost

XGBoost is a gradient boosting algorithm that combines sequentially trained decision trees.

It will be investigated as a boosting-based classification baseline.

**Advantages:**
- Captures nonlinear relationships and feature interactions.
- Supports class-sensitive training configurations.
- Effective for structured feature representations.

**Limitations:**
- Requires hyperparameter tuning.
- May overfit without appropriate regularization.

### 3.3 Deep Learning

#### One-Dimensional Convolutional Neural Network (1D-CNN)

A 1D-CNN learns local patterns directly from ECG waveform segments through convolutional operations.

It will be investigated for automatic extraction of heartbeat morphology.

**Advantages:**
- Learns features directly from raw signal segments.
- Captures local waveform patterns.
- Reduces dependence on manually engineered features.

**Limitations:**
- Requires careful architecture and training configuration.
- May be affected by limited training diversity and class imbalance.

#### Long Short-Term Memory (LSTM)

LSTM is a recurrent neural network designed to learn dependencies in sequential data.

It will be investigated using appropriately constructed ECG sequences, such as consecutive heartbeat or RR interval sequences.

**Advantages:**
- Models temporal dependencies.
- Can capture sequential information.

**Limitations:**
- Requires meaningful sequence construction.
- Can involve higher computational costs.
- May not provide benefits when temporal information is limited.

### 3.4 Hybrid Modelling Strategies

#### Feature-Level Fusion

Feature-level fusion combines complementary ECG representations.

Potential feature groups include:

- Heartbeat morphological characteristics.
- RR interval and temporal characteristics.
- Statistical signal features.

The combined representation will be evaluated using an appropriate traditional machine learning classifier.

The objective is to investigate whether complementary feature information improves classification performance.

#### CNN-LSTM

CNN-LSTM combines convolutional feature extraction with recurrent temporal learning.

The CNN component extracts local waveform characteristics, while the LSTM component models sequential dependencies.

The objective is to investigate whether combining morphological and temporal information provides additional classification benefits.

### 3.5 Model Selection Rationale

The candidate models are selected to represent different learning mechanisms rather than simply selecting popular algorithms.

The investigation will compare:

- Tree-based ensemble learning.
- Margin-based classification.
- Gradient boosting.
- Convolutional feature extraction.
- Sequential learning.
- Feature-level hybrid modelling.

### 3.6 Initial Model Prioritization

**Primary candidates:**
- Random Forest.
- Support Vector Machine.
- XGBoost.
- 1D-CNN.
- Feature-level fusion.

**Secondary candidates:**
- LSTM.
- CNN-LSTM.

The secondary models will be considered after examining the suitability of sequential modelling and computational requirements.

### 3.7 Experimental Considerations

All candidate models will be evaluated using consistent data separation principles.

Feature scaling, feature selection, resampling, and hyperparameter tuning must be performed using training data only.

The test data will remain isolated until final evaluation.

No model is assumed to be superior before experimental comparison.

## 5. Evaluation Protocol and Data Leakage Prevention

### 5.1 Evaluation Strategy

The evaluation strategy will prioritize subject-wise separation to assess whether the proposed modelling approaches can generalize to previously unseen ECG subjects.

A random split of individual heartbeats will not be used as the primary evaluation strategy because heartbeats from the same recording or subject may share similar signal characteristics. Allowing these heartbeats to appear in both training and testing partitions may lead to overly optimistic performance estimates.

### 5.2 Inter-Patient Evaluation

The MIT-BIH Arrhythmia Database contains recordings from multiple subjects. An inter-patient evaluation protocol will be investigated to separate training and testing subjects.

The established DS1/DS2 inter-patient protocol will be considered as a candidate evaluation strategy. Its exact recording assignments and compatibility with the selected experimental design will be verified before implementation.

No subject will be allowed to appear in both the training and testing partitions.

### 5.3 Prevention of Data Leakage

The following operations will be restricted to the training data or training-validation partitions:

- Feature scaling and normalization
- Feature selection
- Missing-value handling, if required
- Oversampling or other class-balancing procedures
- Hyperparameter tuning
- Model selection

The test partition will remain isolated until final evaluation.

### 5.4 Class Imbalance Considerations

ECG heartbeat classes may have substantially different numbers of observations. Therefore, overall accuracy alone will not be sufficient to evaluate model performance.

Class imbalance will be considered during model training and evaluation. Any resampling or class-weighting strategy will be applied only within the training process.

### 5.5 Evaluation Metrics

The following metrics will be considered:

- Accuracy
- Balanced accuracy
- Macro-averaged F1-score
- Per-class precision
- Per-class recall
- Per-class F1-score
- Confusion matrix
- Precision–recall analysis for relevant minority classes

Macro F1-score and balanced accuracy will receive particular attention because they provide a more informative assessment when class distributions are unequal.

### 5.6 Experimental Validity

The evaluation will distinguish between recording-level statistical analysis, heartbeat-level classification, and generalization to unseen subjects.

All model comparisons will use consistent data partitions and evaluation criteria wherever possible.

The final protocol will be confirmed after verifying the subject assignments, annotation mapping, and data preparation requirements.

## 8. Diagnostic-Driven Model Selection Framework

### 8.1 Research Concept

The central contribution of this research is to investigate whether statistical and signal-level characteristics of ECG data can provide useful information for selecting an appropriate machine learning strategy.

Instead of assuming that one model is suitable for every ECG dataset, the proposed framework will examine whether different diagnostic conditions are associated with different modelling requirements.

### 8.2 Diagnostic Information

Potential diagnostic indicators include:

- Stationarity test outcomes
- Rolling mean characteristics
- Rolling variance characteristics
- Signal variability
- Signal quality indicators
- Heartbeat class distribution
- Class imbalance measurements

The diagnostic indicators will be selected based on their methodological justification and availability.

### 8.3 Model Selection Approaches

Two approaches will be investigated.

**Approach A: Global Diagnostic-Driven Selection**

A selection strategy will be developed using diagnostic information and model performance observed within training and validation data.

The objective is to determine whether particular diagnostic conditions are associated with the suitability of particular modelling strategies.

**Approach B: Recording-Specific Selection**

The diagnostic characteristics of an individual recording may be used to select an appropriate modelling strategy.

This approach will require a predefined selection procedure. If applied to test recordings, it will use only information available without accessing their class labels or classification performance.

### 8.4 Candidate Modelling Strategies

The selection framework may consider:

- Traditional machine learning
- Deep learning
- Hybrid feature-level approaches

The final candidate strategies will depend on experimental feasibility and evidence from the model comparisons.

### 8.5 Selection and Evaluation Separation

The diagnostic-driven selection procedure must be developed without using independent test-set labels or performance results.

Model selection, feature selection and hyperparameter tuning will be restricted to training and validation data.

The independent test partition will be used only for final evaluation.

### 8.6 Comparison with Fixed-Model Strategies

The diagnostic-driven strategy will be compared with fixed-model baselines.

The comparison will investigate whether diagnostic information provides useful guidance beyond simply selecting one model based on validation performance.

### 8.7 Feasibility and Limitations

The MIT-BIH database contains a limited number of recordings. Therefore, the reliability of a learned global model-selection system may be restricted by the available recording-level sample size.

The initial study will investigate the feasibility of a simple and interpretable selection strategy before considering more complex approaches.

The findings will be interpreted as initial evidence rather than proof of general applicability across ECG datasets.

## 9. Statistical Comparison and Robustness Assessment

### 9.1 Evaluation Objectives

The statistical evaluation will examine three main aspects:

1. Classification performance across heartbeat categories.
2. Robustness across different recordings and subjects.
3. The additional value of diagnostic-driven model selection compared with fixed-model strategies.

### 9.2 Classification Performance

The following metrics will be reported where appropriate:

- Accuracy
- Balanced accuracy
- Macro-averaged F1-score
- Per-class precision
- Per-class recall
- Per-class F1-score
- Confusion matrix
- Precision–recall analysis for relevant minority classes

Macro F1-score and balanced accuracy will be emphasized because of the potential imbalance between heartbeat categories.

### 9.3 Performance Variability

Model performance will be examined across independent evaluation units.

Where the evaluation design permits, the analysis will report:

- Mean performance
- Standard deviation
- Confidence intervals
- Per-recording or per-subject performance distributions

The aggregation method will be selected according to the final validation protocol.

### 9.4 Statistical Comparison Between Models

Models evaluated on the same independent subjects will be compared using paired statistical procedures when appropriate.

Potential procedures include:

- Wilcoxon signed-rank test
- Paired bootstrap confidence intervals
- Effect-size estimation

The final statistical test will depend on the number of independent evaluation units, the comparison design and the distribution of the paired differences.

Multiple-comparison correction will be considered when several model comparisons are conducted.

### 9.5 Avoiding Pseudoreplication

Individual heartbeats from the same subject or recording will not automatically be treated as independent observations for statistical comparisons.

The analysis will distinguish between:

- Heartbeat-level classification observations
- Recording-level performance summaries
- Subject-level independent evaluation units

This distinction is important for avoiding artificially small p-values and overly optimistic conclusions.

### 9.6 Robustness Assessment

Robustness will be investigated through:

- Performance variation across subjects
- Per-class performance differences
- Sensitivity to class imbalance
- Sensitivity to justified preprocessing choices
- Comparisons across suitable recording conditions

Where feasible, repeated experiments or alternative validation partitions will be considered.

### 9.7 Diagnostic-Driven Model Selection Evaluation

The diagnostic-driven approach will be compared with fixed-model baselines using the same independent evaluation data.

The analysis will investigate whether diagnostic-driven selection provides:

- Improved classification performance
- More consistent performance across subjects
- Better performance for minority heartbeat classes
- Reduced sensitivity to varying recording characteristics

Any observed improvement will be interpreted in relation to uncertainty, effect size and the limitations of the available dataset.

### 9.8 Interpretation and Reporting

Statistical significance will not be interpreted as proof of practical importance.

The final report will distinguish between:

- Statistically significant differences
- Differences in classification performance
- Practical improvements
- Evidence supporting diagnostic-driven model selection
- Limitations arising from dataset size and evaluation design

```python
import wfdb
import pandas as pd
import numpy as np

# Retrieve recording names from the MIT-BIH Arrhythmia Database
record_names = wfdb.get_record_list("mitdb")

print("Total recordings:", len(record_names))
print("\nFirst 10 recording names:")
print(record_names[:10])

# Create a recording inventory
recording_inventory = pd.DataFrame({
    "record_name": record_names
})

display(recording_inventory.head(10))
```

```python
from collections import Counter

# Count annotation symbols across all 48 recordings
symbol_counts = Counter()

for record_name in record_names:
    annotation = wfdb.rdann(
        record_name,
        "atr",
        pn_dir="mitdb"
    )

    symbol_counts.update(annotation.symbol)

# Create a summary table
annotation_summary = pd.DataFrame(
    symbol_counts.items(),
    columns=["annotation_symbol", "count"]
)

annotation_summary = annotation_summary.sort_values(
    by="count",
    ascending=False
).reset_index(drop=True)

print("Total unique annotation symbols:", len(annotation_summary))
print("Total annotation records:", annotation_summary["count"].sum())

display(annotation_summary)
```

```python
# Candidate heartbeat annotation symbols
candidate_beat_symbols = {
    "N", "L", "R", "V", "/", "A", "f", "F",
    "j", "a", "E", "J", "Q", "e", "S"
}

# Identify symbols present in the dataset
observed_symbols = set(symbol_counts.keys())

# Separate candidate heartbeat symbols and other annotations
observed_beat_symbols = observed_symbols.intersection(
    candidate_beat_symbols
)

non_beat_symbols = observed_symbols - candidate_beat_symbols

print("Candidate heartbeat symbols:")
print(sorted(observed_beat_symbols))

print("\nOther annotation symbols:")
print(sorted(non_beat_symbols))

# Display counts for candidate heartbeat symbols
beat_summary = annotation_summary[
    annotation_summary["annotation_symbol"].isin(observed_beat_symbols)
].copy()

display(beat_summary)

print(
    "\nTotal candidate heartbeat annotations:",
    beat_summary["count"].sum()
)
```

```python
# AAMI five-class annotation mapping
aami_mapping = {
    "N": ["N", "L", "R", "e", "j"],
    "S": ["A", "a", "J", "S"],
    "V": ["V", "E"],
    "F": ["F"],
    "Q": ["/", "f", "Q"]
}

# Reverse mapping: annotation symbol -> AAMI class
symbol_to_aami = {
    symbol: aami_class
    for aami_class, symbols in aami_mapping.items()
    for symbol in symbols
}

# Calculate AAMI class distribution
aami_counts = {}

for aami_class, symbols in aami_mapping.items():
    aami_counts[aami_class] = sum(
        symbol_counts.get(symbol, 0)
        for symbol in symbols
    )

aami_distribution = pd.DataFrame({
    "AAMI_class": list(aami_counts.keys()),
    "beat_count": list(aami_counts.values())
})

aami_distribution["percentage"] = (
    aami_distribution["beat_count"]
    / aami_distribution["beat_count"].sum()
) * 100

display(aami_distribution)

print(
    "Total mapped beats:",
    aami_distribution["beat_count"].sum()
)

print(
    "Unmapped candidate symbols:",
    sorted(observed_beat_symbols - set(symbol_to_aami.keys()))
)
```

### Why Conventional Five-Fold Stratified Group Splitting Was Not Selected

Before selecting the final evaluation strategy, we investigated five-fold stratified group cross-validation using subject-level data.

The initial approach assigned each subject to a dominant heartbeat class based on the highest number of beats among the four AAMI classes (N, S, V and F). However, this approach was unsuitable because the dataset was highly imbalanced. Out of 43 subjects, 42 were dominated by the N class, while only one subject was dominated by S.

As a result, the stratification procedure could not adequately distribute the minority classes across the five folds.

The resulting test-fold distributions demonstrated this limitation:

| Fold | N beats | S beats | V beats | F beats |
|---|---:|---:|---:|---:|
| 1 | 19,154 | 391 | 1,365 | 3 |
| 2 | 17,856 | 264 | 1,027 | 11 |
| 3 | 16,953 | 76 | 2,275 | 749 |
| 4 | 17,941 | 1,783 | 1,069 | 29 |
| 5 | 18,221 | 267 | 1,273 | 11 |

The F class was particularly unevenly distributed. Fold 3 contained 749 F beats, whereas the other four folds contained only 54 F beats combined.

This imbalance was strongly influenced by the concentration of F-class examples in two subjects (208 and 213), which together contributed 91.53% of all F beats.

Therefore, the conventional five-fold stratified group approach was not selected for the primary evaluation. Its fold-level class distributions were too inconsistent to support reliable comparisons across folds.

Importantly, this does not mean that five-fold cross-validation is generally unsuitable. The limitation arose from the combination of severe class concentration and the use of each subject's dominant class as the stratification label.

### Alternative Evaluation Strategy: Leave-One-Subject-Out Cross-Validation

To address this limitation, Leave-One-Subject-Out (LOSO) cross-validation was selected for further investigation.

Under this strategy:

- Each subject is held out once for testing.
- The remaining subjects are used for model training.
- All recordings belonging to the same subject remain in the same partition.
- The process is repeated until every subject has been evaluated.

This approach provides a subject-independent evaluation and allows us to examine how model performance varies across individual subjects.

However, LOSO does not eliminate class imbalance or guarantee that every held-out subject contains all four classes. Therefore, class-specific performance, particularly for F, must be interpreted carefully.

The suitability of LOSO will be assessed through a diagnostic inspection of the 43 evaluation folds before model development begins.

```python

from sklearn.model_selection import LeaveOneGroupOut
import numpy as np
import pandas as pd

# Initialize Leave-One-Subject-Out evaluation
logo = LeaveOneGroupOut()

X = np.zeros((len(primary_subject_summary), 1))
groups = primary_subject_summary["subject_id"]

fold_results = []

for fold, (train_idx, test_idx) in enumerate(
    logo.split(X, groups=groups),
    start=1
):
    train_subjects = primary_subject_summary.iloc[train_idx]
    test_subjects = primary_subject_summary.iloc[test_idx]

    fold_results.append({
        "fold": fold,
        "train_subjects": len(train_subjects),
        "test_subject": test_subjects["subject_id"].iloc[0],
        "N": test_subjects["N"].sum(),
        "S": test_subjects["S"].sum(),
        "V": test_subjects["V"].sum(),
        "F": test_subjects["F"].sum()
    })

loso_summary = pd.DataFrame(fold_results)

print("Total LOSO folds:", len(loso_summary))
print("Unique test subjects:", loso_summary["test_subject"].nunique())

print("\nTotal test heartbeat counts:")
display(loso_summary[["N", "S", "V", "F"]].sum())

print("\nSubjects with F beats:", (loso_summary["F"] > 0).sum())
print("Subjects without F beats:", (loso_summary["F"] == 0).sum())

display(loso_summary)
```

The dataset contains 44 eligible recordings from 43 unique subjects. Recordings 201 and 202 belong to the same subject and are therefore treated as a single group.

Under LOSO cross-validation, one subject is held out for testing while the remaining 42 subjects are used for training. This procedure is repeated 43 times, ensuring that every subject is evaluated exactly once.

The evaluation structure was verified programmatically. The checks confirmed 43 unique test subjects, 42 training subjects per fold, and correct grouping of recordings 201 and 202.

This subject-independent design reduces the risk of data leakage caused by placing heartbeats from the same subject in both training and testing partitions.

However, the dataset contains a substantial concentration of F-class examples. Only 16 subjects contain F beats, and two subjects contribute 91.53% of the total F-class examples. Consequently, F-class performance requires careful interpretation, particularly when evaluating generalization to unseen subjects.

```python
from tqdm.auto import tqdm
import numpy as np

pre_samples = 90
post_samples = 90
sampling_rate = 360

heartbeat_segments = []
valid_indices = []

for rec_id in tqdm(
    primary_record_class_df["record_name"].astype(str),
    desc="Extracting ECG segments"
):

    record = wfdb.rdrecord(
        rec_id,
        pn_dir="mitdb",
        physical=True
    )

    signal = record.p_signal[:, 0]  # MLII channel

    record_beats = heartbeat_df[
        heartbeat_df["record_name"] == rec_id
    ]

    for idx, row in record_beats.iterrows():

        center = int(row["sample"])

        start = center - pre_samples
        end = center + post_samples

        # Exclude segments extending beyond recording boundaries
        if start < 0 or end > len(signal):
            continue

        segment = signal[start:end]

        if len(segment) == pre_samples + post_samples:
            heartbeat_segments.append(segment)
            valid_indices.append(idx)

X_ecg = np.asarray(heartbeat_segments, dtype=np.float32)

heartbeat_data = (
    heartbeat_df.loc[valid_indices]
    .reset_index(drop=True)
    .copy()
)

print("Waveform matrix shape:", X_ecg.shape)
print("Heartbeat metadata shape:", heartbeat_data.shape)
print("Expected segment length:", pre_samples + post_samples)
print("Valid heartbeat segments:", len(X_ecg))
print("Excluded boundary segments:", len(heartbeat_df) - len(X_ecg))
```

```python
# Identify consecutive runs of samples at the ADC limits

for rec_id in ["100", "116"]:

    record = wfdb.rdrecord(
        rec_id,
        pn_dir="mitdb",
        physical=True
    )

    signal = record.p_signal[:, 0]
    fs = record.fs

    print(f"\nRecording {rec_id}")
    print("-" * 45)

    for label, mask in [
        ("Minimum limit", signal == signal.min()),
        ("Maximum limit", signal == signal.max())
    ]:

        indices = np.where(mask)[0]

        if len(indices) == 0:
            print(f"{label}: no occurrences")
            continue

        # Divide indices into consecutive runs
        split_positions = np.where(np.diff(indices) > 1)[0] + 1
        runs = np.split(indices, split_positions)

        run_lengths = np.array([len(run) for run in runs])

        print(f"\n{label}")
        print("Number of separate episodes:", len(runs))
        print("Longest episode (samples):", run_lengths.max())
        print("Longest episode (milliseconds):",
              round(run_lengths.max() / fs * 1000, 3))
        print("Episodes with 2 or more samples:",
              np.sum(run_lengths >= 2))
        print("Episodes with 5 or more samples:",
              np.sum(run_lengths >= 5))
```

```python
# Quantify heartbeat segments overlapping saturated samples in recording 116

rec_id = "116"

record = wfdb.rdrecord(
    rec_id,
    pn_dir="mitdb",
    physical=True
)

signal = record.p_signal[:, 0]

# Identify samples at either ADC limit
saturated_mask = (
    (signal == signal.min()) |
    (signal == signal.max())
)

# Get heartbeat annotations from the extracted metadata
rec_beats = heartbeat_data[
    heartbeat_data["record_name"].astype(str) == rec_id
].copy()

# Each segment contains 180 samples, centered on its annotation
half_window = 90

overlap_counts = []

for _, beat in rec_beats.iterrows():

    center = int(beat["sample"])
    start = center - half_window
    end = center + half_window

    start = max(0, start)
    end = min(len(signal), end)

    saturated_samples = saturated_mask[start:end].sum()

    overlap_counts.append(saturated_samples)

rec_beats["saturated_samples"] = overlap_counts
rec_beats["saturation_percentage"] = (
    rec_beats["saturated_samples"] / 180 * 100
)

affected_beats = rec_beats[
    rec_beats["saturated_samples"] > 0
]

print("Recording:", rec_id)
print("Total extracted heartbeats:", len(rec_beats))
print("Heartbeats overlapping saturation:", len(affected_beats))
print(
    "Percentage affected:",
    round(len(affected_beats) / len(rec_beats) * 100, 3),
    "%"
)

print("\nAffected heartbeats by AAMI class:")
display(
    affected_beats["aami_class"]
    .value_counts()
    .to_frame("affected_beats")
)

print("\nMost affected heartbeat segments:")
display(
    affected_beats[
        [
            "sample",
            "annotation_symbol",
            "aami_class",
            "saturated_samples",
            "saturation_percentage"
        ]
    ].sort_values(
        "saturated_samples",
        ascending=False
    ).head(15)
)
```

# Notebook 02 — Dataset Preparation and Experimental Design

## 1. Purpose of This Notebook

The purpose of this notebook is to develop and evaluate different machine-learning strategies for ECG heartbeat classification.

This study follows a diagnostic-driven approach. Instead of simply comparing several classifiers, we aim to investigate whether statistical and signal-level characteristics of ECG recordings can help us select more suitable modelling strategies.

The main research question is:

*Can statistical and signal-level characteristics help us select a more suitable modelling strategy for different ECG recording conditions?*

The modelling experiments will compare traditional machine learning, deep learning, and hybrid approaches.

The main experimental stages are:

1. Prepare the ECG heartbeat dataset.
2. Extract meaningful features from individual heartbeat segments.
3. Develop traditional machine-learning models.
4. Develop deep-learning models.
5. Investigate hybrid modelling strategies.
6. Compare fixed-model approaches with diagnostic-driven model selection.
7. Evaluate model performance under subject-independent conditions.

---

## 2. Dataset Description

### 2.1 MIT-BIH Arrhythmia Database

The dataset used in this study is the **MIT-BIH Arrhythmia Database**, obtained through the WFDB Python library.

The database contains ECG recordings with expert annotations identifying different heartbeat types.

| Property | Description |
|---|---|
| Dataset | MIT-BIH Arrhythmia Database |
| Original recordings | 48 |
| Number of subjects | 47 |
| Sampling frequency | 360 Hz |
| ECG channels | 2 |
| Annotation type | Expert-annotated heartbeat labels |
| Signal unit | Millivolts (mV) |

Each recording contains ECG signals and corresponding annotations. These annotations provide the reference labels used to identify heartbeat classes.

### 2.2 Subject Identification

The database contains 48 recordings from 47 subjects.

Recordings 201 and 202 belong to the same subject. Therefore, they must be assigned the same subject identifier during subject-independent evaluation.

We created a recording-to-subject mapping to maintain this relationship.

This is important because recordings from the same individual should not be treated as independent subjects when evaluating model generalization.

---

## 3. Selection of Recordings

We examined the conventional DS1 and DS2 recording lists commonly used in MIT-BIH heartbeat classification experiments.

The initial recording lists contained:

- DS1: 22 recordings.
- DS2: 22 recordings.
- Total: 44 recordings.

Four recordings were not included in these lists:

- 102
- 104
- 107
- 217

We also identified a subject-overlap issue: recordings 201 and 202 belong to the same subject, but they appear in different groups.

Consequently, the conventional DS1/DS2 assignment cannot automatically be considered fully subject-independent under our subject mapping.

For the current experimental dataset, we selected the 44 recordings and retained the subject identifiers separately.

**Important methodological consideration:** The final evaluation protocol and the exact relationship to published DS1/DS2 benchmarks must be verified before reporting benchmark comparisons. We will use subject identifiers to prevent subject leakage in our own evaluation.

---

## 4. Heartbeat Classification and AAMI Mapping

We used a provisional AAMI-based classification mapping to group the original MIT-BIH annotation symbols.

| AAMI class | Annotation symbols | Description |
|---|---|---|
| N | N, L, R, e, j | Normal and bundle branch-related beats |
| S | A, a, J, S | Supraventricular ectopic beats |
| V | V, E | Ventricular ectopic beats |
| F | F | Fusion beats |
| Q | /, f, Q | Paced and unclassifiable beats |

The original annotation symbols were mapped to their corresponding AAMI groups.

### 4.1 Selection of Four Primary Classes

Although the original mapping contains five classes, we selected four classes for the initial modelling experiments:

- N
- S
- V
- F

The Q class was not included in the initial four-class modelling dataset.

The reason is that Q contains a large proportion of paced beats, making it a heterogeneous group compared with the other classes.

The Q class contained 8,043 annotations in the earlier 48-recording analysis:

| Q annotation symbol | Count |
|---|---:|

## 6. ECG Heartbeat Segmentation

After extracting the heartbeat annotation positions, we extracted fixed-length ECG segments around each annotated heartbeat.

The MLII channel (first ECG channel) was selected for the initial experiment.

### 6.1 Segmentation Configuration

The segmentation window was defined as:

- 90 samples before the annotation.
- 90 samples after the annotation.
- Total segment length: 180 samples.

At a sampling frequency of 360 Hz:

\[
\text{Segment duration} = \frac{180}{360}=0.5 \text{ seconds}
\]

Therefore, each heartbeat segment represents approximately 0.5 seconds of ECG signal.

### 6.2 Boundary Handling

Some annotations were located too close to the beginning or end of a recording to extract a complete 180-sample segment.

We excluded these incomplete segments.

The extraction results were:

| Property | Result |
|---|---:|
| Initial heartbeat annotations | 100,718 |
| Extracted heartbeat segments | 100,693 |
| Excluded boundary segments | 25 |
| Samples per segment | 180 |
| ECG channel | MLII |

The resulting ECG matrix was:

\[
X_{\text{ecg}} \in \mathbb{R}^{100693 \times 180}
\]

Each row represents one heartbeat, and each column represents a time point within that heartbeat.

The corresponding metadata table contains the recording, subject, annotation position, original symbol, and AAMI class.

## 8. Investigation of Within-Window Baseline Differences

We also investigated possible amplitude shifts within heartbeat segments.

For each segment, we calculated the mean of the first 30 samples and the mean of the last 30 samples.

The difference was defined as:

\[
D_{\text{edge}} =
\overline{x}_{\text{last 30}} -
\overline{x}_{\text{first 30}}
\]

This measure was used as an exploratory indicator of differences between the two edges of a heartbeat window.

The analysis showed variation between recordings.

However, the first and last 30 samples may contain neighbouring cardiac activity. Therefore, this measure cannot be treated as a direct measurement of baseline wander.

We refer to it as **within-window baseline difference**, not baseline drift.

This distinction is important when interpreting the results.

## 9. Comparison of Three Signal Representations

We created three versions of the extracted ECG segments to investigate the effects of simple amplitude preprocessing.

The original data were preserved.

### 9.1 P0 — Raw ECG

The original heartbeat segments were retained without amplitude correction.

\[
X_{\text{raw}} = X_{\text{ecg}}
\]

### 9.2 P1 — Mean-Centered ECG

For each heartbeat, we subtracted its own mean amplitude.

\[
X_{\text{mean-centered}} =
X_{\text{ecg}} -
\operatorname{mean}(X_{\text{ecg}})
\]

This removes the mean offset within each segment.

### 9.3 P2 — Edge-Corrected ECG

We calculated the average amplitude of the first and last 30 samples and subtracted this value from the entire heartbeat segment.

\[
X_{\text{edge-corrected}} =
X_{\text{ecg}} - B_{\text{edge}}
\]

where \(B_{\text{edge}}\) is the average of the two edge regions.

This is an exploratory preprocessing method. It has not yet been established as a clinically or physiologically appropriate baseline correction method.

### 9.4 Comparison of Signal Representations

| Statistic | P0: Raw | P1: Mean-centered | P2: Edge-corrected |
|---|---:|---:|---:|
| Overall mean | -0.32471 | 0.00000 | 0.03531 |
| Overall standard deviation | 0.50314 | 0.40066 | 0.42166 |
| Minimum | -5.12000 | -3.37314 | -3.94950 |
| Maximum | 5.11500 | 6.76528 | 5.82442 |

The preprocessing methods changed the amplitude distributions.

In particular, mean-centering and edge correction sometimes increased the maximum absolute amplitude.

Therefore, preprocessing should not automatically be assumed to improve signal quality.

We will evaluate the different representations experimentally rather than selecting one based only on descriptive statistics.

## 11. Investigation of Recording 116

During the extreme-amplitude investigation, we identified an unusual amplitude pattern in recording 116.

- Two ECG channels.
- MLII as the first channel.

- ADC gain of 200.

## 16. Important Methodological Decisions Before Model Training

Several decisions still require attention during the next stages.

### 16.1 Subject-independent evaluation

Heartbeat segments from the same subject must not appear in both training and testing data.

We will use subject identifiers when creating the evaluation splits.

The previously examined Leave-One-Subject-Out (LOSO) structure contains 43 folds, with one subject held out in each fold.

This provides a possible subject-independent evaluation strategy.

### 16.2 Class imbalance

The four classes are highly imbalanced, particularly F.

We will investigate appropriate approaches such as:

- Class-weighted learning.
- Training-only oversampling.
- Other suitable imbalance-handling techniques.

Any resampling, scaling, feature selection, and model tuning must be performed using training data only.

### 16.3 Preprocessing comparison

We have created three signal representations:

- Raw ECG.
- Mean-centered ECG.
- Edge-corrected ECG.

Their effects on model performance have not yet been tested.

We will compare them experimentally rather than assuming that one is superior.

### 16.4 Evaluation metrics

Accuracy alone will not be sufficient because of the class imbalance.

The main evaluation metrics will include:

- Macro-F1 score.
- Balanced accuracy.
- Per-class precision.
- Per-class recall.
- Per-class F1-score.
- Confusion matrix.
- Precision-recall analysis for minority classes.

### 16.5 Diagnostic-driven model selection

The central contribution of this research is not simply to identify the best classifier.

We aim to investigate whether recording-level statistical and signal-level diagnostics can help determine which modelling strategy is more suitable under different recording conditions.

This will require a direct experimental comparison between:

1. Fixed-model approaches.
2. Diagnostic-guided model selection.

The diagnostic information must be calculated without using test labels or information from the held-out subject.
